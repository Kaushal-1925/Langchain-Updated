# 🚀 Integrating Qwen 3 (32B) on Groq Cloud

This notebook demonstrates how to integrate and run the highly performant **Qwen 3 (32B)** model on **Groq** using LangChain's `ChatGroq` integration.

### Prerequisites
1. Ensure you have the `langchain-groq` package installed.
2. Make sure your `GROQ_API_KEY` is defined in your `.env` file.

```python
import os
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_groq import ChatGroq

# 1. Load the environment variables from your .env file
load_dotenv()

# 2. Initialize the Qwen 3 (32B) model on Groq
llm = ChatGroq(
    model="qwen/qwen3-32b",
    temperature=0.7,
)

# 3. Prepare the conversation messages
messages = [
    SystemMessage(content="You are an expert, encouraging AI coding assistant."),
    HumanMessage(content="Write a clean, optimized Python function to reverse a string and explain it in one sentence.")
]

# 4. Invoke the model
print("📡 Sending request to Groq (model: qwen/qwen3-32b)...\n")
response = llm.invoke(messages)

# 5. Display the response
print("✨ Response:")
print(response.content)
```
