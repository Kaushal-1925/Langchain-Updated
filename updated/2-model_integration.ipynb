```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:qwen/qwen3-32b")
response = model.invoke("Why do parrots talk?")
response
```

### Streaming 

Most models can stream their output content while it is being generated. By dispaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real time

```python
response = model.invoke("Write me a 200 word paragraph on Artifical Intelligence")
response.content
```

```python
for chunk in model.stream("Write me a 200 word paragraph on Artifical Intelligence"):
    print(chunk.text, end="|", flush=True)
```

```python
for chunk in model.stream("Who is elon musk?"):
    print(chunk.text, end="|", flush=True)
```

### Batch

Batching a collection of independent requests to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:

```python
from langchain_core.runnables import config
responses = model.batch(
    ["why do parrots have colourful feathers?",
    "How do airplanes fly",
    "Are both of the things connected"
    ],
    config = {
        'max_concurrency' : 5, # Limit to 4 parallel calls
    }
)
for response in responses:
    print(response)
```
